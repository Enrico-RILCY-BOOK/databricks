```python
from pyspark.sql import functions as F

#1. Construire un jeu de données de 20 Millions de tickets. Rien ne s'exécute encore
tickets = (spark.range(0,20000000)
               .withColumn("magasin", (F.col("id") % 800).cast("int"))
               .withColumn("montant", F.round(F.rand(42)*100, 2)))

#2. Transformation Narrow. Pas de shuffle
gros = tickets.filter(F.col("montant") > 50)

#3. Transformation Wide. Shuffle
ca = gros.groupBy("magasin").agg(F.sum("montant").alias("ca"))

#4. Lire le plan avan exécution
ca.explain()

#5. Lancer l'exécution
ca.orderBy(F.col("ca").desc()).show(5)
```

```python
from pyspark.sql import functions as F

#1. Construire un jeu de données de 20 Millions de tickets. Rien ne s'exécute encore
tickets = (spark.range(0,20000000)
               .withColumn("magasin", (F.col("id") % 800).cast("int"))
               .withColumn("montant", F.round(F.rand(42)*100, 2)))

#2. Transformation Narrow. Mais ajout d'une répartition donc shuffle
gros = tickets.repartition(8).filter(F.col("montant") > 50)

#3. Sélection du résultat
ca = gros.select("magasin", "montant")

#4. Lire le plan avan exécution
ca.explain()

#5. Lancer l'exécution
ca.orderBy(F.col("montant").desc()).show(5)
```

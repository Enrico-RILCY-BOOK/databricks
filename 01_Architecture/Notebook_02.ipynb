```sql
%sql
-- Requete 1 : filtre + agregation
EXPLAIN FORMATTED
SELECT o_orderpriority, count(*) AS nb
FROM samples.tpch.orders
WHERE o_orderdate >= '1995-01-01'
GROUP BY o_orderpriority;
```

```sql
-- Requete 2 : jointure + agregation
EXPLAIN FORMATTED
SELECT c.c_mktsegment, sum(o.o_totalprice) AS ca
FROM samples.tpch.orders o
JOIN samples.tpch.customer c ON o.o_custkey = c.c_custkey
GROUP BY c.c_mktsegment;
```

```sql
-- Requete 3 : fonction de fenetre
EXPLAIN FORMATTED
SELECT o_custkey, o_orderdate, o_totalprice,
       row_number() OVER (PARTITION BY o_custkey ORDER BY o_totalprice DESC) AS rang
FROM samples.tpch.orders;
```
